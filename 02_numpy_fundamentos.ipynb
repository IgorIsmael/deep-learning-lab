## 1. Importação

Por convenção, importamos NumPy como `np`. Todos os exemplos deste notebook usarão arrays NumPy.


```python
import numpy as np
```

## 2. Escalar, vetor, matriz e tensor

- **Escalar:** um único valor, sem eixos;
- **Vetor:** uma sequência em um eixo;
- **Matriz:** linhas e colunas;
- **Tensor:** termo geral para arrays com qualquer número de eixos.

No contexto de Deep Learning, “tensor” também é usado de forma ampla para representar os dados processados pela rede.


```python
escalar = np.array(5)

vetor = np.array([1, 2, 3])

matriz = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

tensor = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])
```

### Visualizando os valores

A impressão permite observar como o NumPy representa cada estrutura. Os colchetes aninhados indicam o número de eixos.


```python
print("Escalar:")
print(escalar)

print("\nVetor:")
print(vetor)

print("\nMatriz:")
print(matriz)

print("\nTensor:")
print(tensor)
```

## 3. Número de dimensões com `ndim`

`ndim` informa quantos eixos o array possui. Isso não é a mesma coisa que a quantidade total de elementos.


```python
print("Dimensão do escalar:", escalar.ndim)
print("Dimensão do vetor:", vetor.ndim)
print("Dimensão da matriz:", matriz.ndim)
print("Dimensão do tensor:", tensor.ndim)
```

## 4. Formato com `shape`

`shape` informa o tamanho de cada eixo. Exemplos:

- `(3,)`: vetor com três elementos;
- `(2, 3)`: matriz com duas linhas e três colunas;
- `(2, 2, 2)`: tensor com três eixos, cada um de tamanho dois.


```python
print("Shape do escalar:", escalar.shape)
print("Shape do vetor:", vetor.shape)
print("Shape da matriz:", matriz.shape)
print("Shape do tensor:", tensor.shape)
```

## 5. Representando um dataset

Cada linha representará uma pessoa e cada coluna uma característica. Podemos interpretar as colunas como idade, renda e quantidade de dependentes.


```python
dados = np.array([
    [25, 3000, 2],
    [30, 5000, 5],
    [22, 2000, 1],
    [40, 7000, 10]
])

print(dados)
print("Shape:", dados.shape)
```

### Exemplos e características

Em Machine Learning, uma matriz de entrada costuma seguir o formato:

`(número de exemplos, número de características)`

Assim, `shape[0]` acessa as linhas e `shape[1]` acessa as colunas.


```python
print("Número de exemplos:", dados.shape[0])
print("Número de características:", dados.shape[1])
```

## 6. Revisão integrada

O bloco abaixo reúne a criação, a impressão e a interpretação do dataset. Ele repete parte do exemplo anterior de propósito, funcionando como uma revisão executável.


```python
dados = np.array([
    [25, 3000, 2],
    [30, 5000, 5],
    [22, 2000, 1],
    [40, 7000, 10]
])

print("Dados:")
print(dados)

print("Shape:", dados.shape)

print("Número de exemplos:", dados.shape[0])
print("Número de características:", dados.shape[1])
```

## 🧠 Conclusão

Neste notebook, aprendemos a ler o formato dos dados antes de realizar qualquer cálculo. Essa habilidade evita erros de incompatibilidade dimensional em redes neurais.
